**BM25**

```python
import math
docs = {"D1": "scrum training für teams".split(),
        "D2": "konflikt training für vertrieb und teams".split(),
        "D3": "excel training".split()}
N, avg = len(docs), sum(len(d) for d in docs.values()) / len(docs)
k1, b = 1.5, 0.75

def idf(word):
    n = sum(word in d for d in docs.values())
    return math.log(1 + (N - n + 0.5) / (n + 0.5))

def bm25(query, doc):
    score = 0
    for w in query.split():
        tf = doc.count(w)
        if tf:
            score += idf(w) * tf * (k1 + 1) / (tf + k1 * (1 - b + b * len(doc) / avg))
    return score

for name, d in docs.items():
    print(name, round(bm25("scrum training", d), 3))
```

Here D1 has the highest score, hence the closest profile match (only using BM25 in search)

**Where BM25 Fails**

*   Synonyms
*   Filler words (noise)
*   Other language
*   Compounds
*   Problem described, no actual word
